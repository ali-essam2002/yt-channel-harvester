```python
import pandas as pd

INPUT_FILE = "output\AndroidBasha_videos_20261021_124244.csv"
OUTPUT_FILE = "output\AndroidBasha_videos_cleaned.csv"

df = pd.read_csv(INPUT_FILE, encoding="utf-8-sig")
print(f"Rows before cleaning: {len(df)}")

# Strip whitespace from text cells, then treat empty strings as nulls
text_cols = df.select_dtypes(include="object").columns
for col in text_cols:
    df[col] = df[col].str.strip()
df = df.replace(r"^\s*$", pd.NA, regex=True)


df = df.dropna(how="all")
df = df.dropna(how="any")



df = df.reset_index(drop=True)
print(f"Rows after cleaning: {len(df)}")

# utf-8-sig keeps Arabic titles readable when opened in Excel
df.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")
print(f"Saved to {OUTPUT_FILE}")
```

```python
import logging
from pathlib import Path

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger(__name__)

file_size_kb = Path(OUTPUT_FILE).stat().st_size / 1024
logger.info(f"Saved {len(df)} rows to {OUTPUT_FILE}  ({file_size_kb:.1f} KB)")

print(f"\n{'─'*55}")
print(f"  Reloaded shape : {df.shape}")
print(f"  File    : {OUTPUT_FILE}")
print(f"  Rows    : {len(df):,}")
print(f"  Columns : {len(df.columns)}")
print(f"  Size    : {file_size_kb:.1f} KB")
print("  Head    :")
print(df.head(5))
print(f"{'─'*55}")
```
